Исходный массив:

```python
nums = [1,5,3,2,2,7,6,4,8,9]
```

#### Решение 1: полтора часа страданий и тупости
1. Сортируем список:

```python

sorted_nums = sorted(nums)
```

2. Проверяем исключительные случаи \[1,1] и \[2,2]:

```python
if len(nums) == 2:
    if nums[0] == 1:
        ans = [1, 2]
    else:
        ans = [2, 1]
        
    print(ans)
```

3. Находим повторяющееся значение и записываем в ans\[0]:

```python
last_num = 0
ans = []

for num in sorted_nums:
    if num == last_num:
        ans.append(num)
    last_num = num

print(ans[0])
```

4. Убираем повторяющееся значение из массива и проходимся по нему в поисках отсутствующего значения:

```python
set_nums = sorted_nums.copy()
set_nums.remove(ans[0])

for i in range(len(set_nums)):
    print(f"{i}й элемент сета: {set_nums[i]}")

    if set_nums[i] != i + 1:
        ans.append(i + 1)
        print(ans)
```

5. Если предыдущий перебр не дал результата - отсутствующее значение - крайнее число массива:

```python
ans.append(len(nums))
print(ans)
```

#### Решение 2: лучшее читаемое
1. Создаем массив нулей длинны n + 1, чтобы посчитать количество вхождений элементов в исходный массив. n + 1 нужен для того, чтобы индексы массива вхождений соответствовали элементу исходного списка:

```python
n = len(nums)
frequency = [0] * (n + 1)
```

2. Находим число вхождений элементов:

```python
for num in nums:
    frequency[num] += 1
```

3. Проходимся по массиву вхождений и ищем повторяющийся и отсутствующий элемент:

```python
duplicate = 0
missing = 0
for i in range(1, n + 1):
    if frequency[i] == 2:
        duplicate = i
    if frequency[i] == 0:
        missing = i

print([duplicate, missing])
```
